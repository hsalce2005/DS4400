### Part C

```python
# genrate 3 random 3x3 matrices with integers from -10 to 10
matrices = []
for i in range(3):
    M = np.random.randint(-10, 11, size=(3, 3))
    matrices.append(M)
```

```python
 # print each matrix out
for i, M in enumerate(matrices):
    print(f"Matrix {i+1}:")
    print(M)
    
    # compute determinant to determine if invertible
    det = np.linalg.det(M)
    print(f"Determinant: {np.round(det, 4)}")
    
    if abs(det) < 0.000001:
        print("Matrix is singular so it is not invertible")
        
    else:
        # compute inverse using Numpy
        M_inv = np.linalg.inv(M)
        print(f"\nInverse of Matrix {i+1}:")
        print(np.round(M_inv, 4))
        
        # M * M_inverse should equal Identity
        product = M @ M_inv
        print(f"\nVerification of Identity:")
        print(np.round(product, 4))
        print()
```
